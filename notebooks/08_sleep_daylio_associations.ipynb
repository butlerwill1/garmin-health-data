# Sleep–Daylio associations

Daylio entries are joined to the same morning date as the sleep end time. Medication results are observational and vulnerable to confounding by indication; they do not establish that medication caused an outcome.

```python
from garmin_daylio_analysis.sleep import load_local_sleep_analysis, sleep_tag_associations

sleep = load_local_sleep_analysis()
nights = sleep.nights
tags = [column for column in nights if column.startswith("tag_")]
len(tags), tags
```

```python
same_date = sleep_tag_associations(nights)
same_date.sort_values(["fdr_q_value", "p_value"], na_position="last").head(30)
```

```python
# Sensitivity check: tags recorded on the previous calendar day, for entries made at night.
prior_tags = [column for column in nights if column.startswith("prior_day_tag_")]
prior_day = sleep_tag_associations(nights, prior_tags)
prior_day.sort_values(["fdr_q_value", "p_value"], na_position="last").head(30)
```

```python
focus = ["Middle of Night Wake", "Early wake", "Sleeping medication", "Organic sleep pills", "Restless", "tired", "sleep 8 hours"]
same_date[same_date.tag.isin(focus)].sort_values(["tag", "metric"]) 
```
