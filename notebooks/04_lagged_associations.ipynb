# Lagged associations

Positive lag means Garmin(t-lag) is paired with mood(t). This is directional timing, not proof of causation.

```python
import seaborn as sns
from garmin_daylio_analysis.statistics import lagged_associations
from garmin_daylio_analysis.workflow import load_local_analysis

config, analysis_daily = load_local_analysis()
lag_results = lagged_associations(
    analysis_daily, ['steps', 'resting_heart_rate', 'stress_average'],
    max_lag_days=config.max_lag_days, min_observations=config.min_observations,
)
sns.lineplot(data=lag_results.query("status == 'tested'"), x='lag_days', y='effect_size', hue='feature', marker='o')
```
