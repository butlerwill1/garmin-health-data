# Adjusted and robustness analysis

The model adjusts for weekday, long-term trend, and annual seasonality. Compare the daily mean outcome against the median and final check-in before treating a finding as robust.

```python
import pandas as pd
from garmin_daylio_analysis.statistics import adjusted_ordinal_association
from garmin_daylio_analysis.synthetic import make_synthetic_analysis_daily

analysis_daily = make_synthetic_analysis_daily()
features = ['steps', 'resting_heart_rate', 'stress_average']
outcomes = ['mood_mean', 'mood_median', 'mood_last']
adjusted = pd.DataFrame([
    {**adjusted_ordinal_association(analysis_daily, feature, outcome, bootstrap_resamples=20 if outcome == 'mood_mean' else 0), 'outcome': outcome}
    for outcome in outcomes for feature in features
])
display(adjusted)
```
