# Exploratory trends

Plots describe patterns and associations; they do not establish causes.

```python
import matplotlib.pyplot as plt
import seaborn as sns
from garmin_daylio_analysis.synthetic import make_synthetic_analysis_daily

analysis_daily = make_synthetic_analysis_daily()
fig, axes = plt.subplots(1, 2, figsize=(12, 4))
sns.countplot(data=analysis_daily, x='mood_mean', ax=axes[0])
axes[0].set(title='Daily mean mood', xlabel='Mood score')
sns.boxplot(data=analysis_daily, x='mood_mean', y='steps', ax=axes[1])
axes[1].set(title='Steps by daily mood', xlabel='Mood score')
fig.tight_layout()
```
